```python
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
```

```python
import os
os.environ['hf_key']='key'
```

```python
llm=HuggingFaceEndpoint(
     repo_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
        #task="conversational",
        provider="featherless-ai"
)
```

```python
model=ChatHuggingFace(
    llm=llm
)
```

```python
response=model.invoke("who is the prime minister  of india?")
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate,PromptTemplate
```

```python
prompt1=PromptTemplate(
    template='write a detailed report on {topic}',
    input_variables=['topic']
)
```

```python
prompt2=PromptTemplate(
    template='Write a 5 line summmary on the folliwing text. /n {text}',
    input_variables=['text']
)
```

```python
template1=prompt1.invoke({'topic':'blackhole'})
```

```python
result=model.invoke(template1)
```

```python
template2=prompt2.invoke({'text':result.content})
```

```python
result1=model.invoke(template2)
```

```python
print(result1.content)
```
