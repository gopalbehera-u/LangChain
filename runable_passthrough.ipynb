```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence,RunnableParallel,RunnablePassthrough
```

```python
psrser=StrOutputParser()
model=ChatOpenAI(model='gpt-5.4-mini')
import os 
os.environ['OPENAI_API_KEY']="your key"
```

```python
passthrough=RunnablePassthrough()
```

```python
print(passthrough.invoke("name"))
```

```python
prompt1=PromptTemplate(
    template="tell me a joke about {topic}",
    input_variables=['topic']
)
prompt2=PromptTemplate(
    template="explain the following joke - {topic}",
    input_variables=['topic']
)
joke_gen=RunnableSequence(prompt1,model,psrser)
```

```python
parallel_chaain=RunnableParallel({
    "joke":RunnablePassthrough(),
    'explanation':RunnableSequence(prompt2,model,psrser)
})
```

```python
final_chain=RunnableSequence(joke_gen,parallel_chaain)
```

```python
print(final_chain.invoke({'topic':'ms dhoni'}))
```
