```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
retriver=WikipediaRetriever(top_k_results=2,lang='en')
```

```python
query="narendra modi"
```

```python
docs=retriver.invoke(query)
```

```python
for i , doc in enumerate(docs):
    print(f"result {i+1}")
    print(f"content {doc.page_content}")
```
