```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence,RunnableParallel,RunnablePassthrough,RunnableLambda
```

```python
psrser=StrOutputParser()
model=ChatOpenAI(model='gpt-5.4-mini')
import os 
os.environ['OPENAI_API_KEY']="your key"
```

```python
def word_counter(text):
    return len(text.split())

runnable_word_counter=RunnableLambda(word_counter)

print(runnable_word_counter.invoke('hi there how are you '))
```

```python
prompt1=PromptTemplate(
    template="write a dark joke about {topic}",
    input_variables=['topic']
)
```

```python
joke_gen=RunnableSequence(prompt1,model,psrser)
```

```python
paralle_chain=RunnableParallel({
    'joke':RunnablePassthrough(),
    'word count':RunnableLambda(word_counter)
})
```

```python
final_chain=RunnableSequence(joke_gen,paralle_chain)
```

```python
result=final_chain.invoke({'topic':'virat kohli'})
```

```python
final_result="""{} \n word count - {}""".format(result['joke'],result['word count'])
```
